# prism

prism is a logical-physical boundary compiler for dual-rail qubits. Each logical gate carries a set of physical realizations, and the router picks a realization for each gate during routing instead of fixing one up front.

This notebook compiles a 4-qubit GHZ circuit onto a 2x4 grid device (8 physical rails, 2 per logical qubit).

```python
import importlib.util

if importlib.util.find_spec("prism") is None:
    %pip install prism-qc
```

```python
from qiskit import QuantumCircuit
from qiskit.transpiler import CouplingMap

from prism.encodings import DualRail
from prism.pipeline import build_pm

qc = QuantumCircuit(4)
qc.h(0)
for i in range(3):
    qc.cx(i, i + 1)
qc.draw("mpl")
```

```python
pm = build_pm(CouplingMap.from_grid(2, 4), enc=DualRail, seed=42)
out = pm.run(qc)
print(pm.property_set["circuit_metrics"])
```

```python
out.draw("mpl", fold=-1)
```
